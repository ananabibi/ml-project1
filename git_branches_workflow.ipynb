# Git / GitHub — workflow simple avec des branches

**Règle d'or :** on ne code pas directement sur `main`. Chacun crée une branche pour sa tâche.

## 1. Partir d'un `main` à jour

````python
git switch main
git pull
````

## 2. Créer sa branche

Exemple : tu travailles sur l'authentification de l'API.

````python
git switch -c api-auth
````

Si vous êtes 3 sur l'API, vous pouvez avoir par exemple : `api-auth`, `api-users`, `api-orders`.

## 3. Coder puis commit

````python
git add .
git commit -m "ajoute l'authentification"
````

## 4. Push la branche sur GitHub

La première fois :

````python
git push -u origin api-auth
````

Les fois suivantes sur la même branche :

````python
git push
````

## 5. Faire une Pull Request sur GitHub

Branche → `main` → vérifier → merge.

## 6. Les autres récupèrent le nouveau `main` dans leur branche

````python
git switch main
git pull
git switch api-users
git merge main
````

Remplace `api-users` par le nom de ta branche.

## 7. Routine complète

````python
# 1. Mettre main à jour
git switch main
git pull

# 2. Créer une branche
git switch -c nom-de-ma-branche

# 3. Je code...

# 4. Commit
git add .
git commit -m "description de mes changements"

# 5. Push
git push -u origin nom-de-ma-branche

# 6. Pull Request sur GitHub puis merge dans main
````

## 8. Commandes utiles

````python
# Voir les branches
git branch

# Changer de branche
git switch nom-de-branche

# Voir l'état des fichiers
git status

# Voir les commits
git log --oneline --graph --all
````

## 9. Règles d'or

- Ne pas coder directement sur `main`.
- 1 branche = 1 tâche / sous-tâche.
- Faire des petits commits réguliers.
- Faire `git pull` sur `main` avant de créer une nouvelle branche.
- Après qu'une autre branche est mergée dans `main`, récupérer `main` dans sa branche.
- Éviter que plusieurs personnes modifient les mêmes lignes pendant longtemps.

## Exemple à 3 personnes

```text
main
├── api-auth      → personne 1
├── api-users     → personne 2
└── api-orders    → personne 3
```

Quand une personne finit : branche → Pull Request → merge dans `main`. Les autres récupèrent ensuite `main` dans leur branche.
